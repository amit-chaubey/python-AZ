```python
import requests
```

```python
url = "https://api.github.com/search/repositories?q=language:typscripts+sort:stars+stars:>10000"
```

```python
header = {"Accept": "application/vnd.github.v3+json"}

response = requests.get(url, headers=header)

print("status code: ", response.status_code)
```

```python
response_dict = response.json()

print(response_dict.keys())
```

```python
print("Total repositories: ", response_dict["total_count"])

repo_dicts = response_dict["items"]
print("Repositories returned: ", len(repo_dicts))


repo_dict = repo_dicts[0]
print("\nKeys: ", len(repo_dict))
[key for key in sorted(repo_dict.keys())]
    

```

```python
def first_repo_info(repo_dict):
    print("\nSelected information about first repository:")
    print("Name: ", repo_dict["name"])
    print("Owner: ", repo_dict["owner"]["login"])
    print("Stars: ", repo_dict["stargazers_count"])
    print("Repository: ", repo_dict["html_url"])
    print("Created: ", repo_dict["created_at"])
    print("Updated: ", repo_dict["updated_at"])
    print("Description: ", repo_dict["description"])
    
```

```python
first_repo_info(repo_dict)
```

```python
def each_repo_info(repo_dicts):
    # Create list of formatted repo info using list comprehension
    repo_info_list = [
        (
            f"\nSelected information about each repository:\n"
            f"Name: {repo['name']}\n"
            f"Owner: {repo['owner']['login']}\n"
            f"Stars: {repo['stargazers_count']}\n"
            f"Repository: {repo['html_url']}\n"
            f"Created: {repo['created_at']}\n"
            f"Updated: {repo['updated_at']}\n"
            f"Description: {repo['description']}"
        )
        for repo in repo_dicts
    ]
    
    # Print results
    for info in repo_info_list:
        print(info)

    return repo_info_list  # optional if you want to use the data later


# call the function
each_repo_info(repo_dicts)

```

```python
import plotly.express as px

repo_names = [repo["name"] for repo in repo_dicts]
stars = [repo["stargazers_count"] for repo in repo_dicts]

fig = px.bar(x=repo_names, y=stars, title="GitHub Stars per Repository",
             labels={"x": "Repository", "y": "Stars"})
fig.show()
```
